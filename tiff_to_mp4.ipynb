```python
import subprocess, tempfile, os
from pathlib import Path

FRAME_DIR = Path(r"F:\Yichen\phantom_videos_backup\DNg27_CsChR\2026_0331_144211_DNg27_CsChr_R13D04_CsChR_7d_F_Fly3_Trial3_0ms_thorax_sp2_oscillating_sine_wave_40_ON_PhantomCamera")
OUTPUT = FRAME_DIR.with_suffix(".mp4")
FPS    = 300

files = sorted(FRAME_DIR.glob("*.tif"))
print("frames:", len(files))

with tempfile.NamedTemporaryFile("w", suffix=".txt", delete=False, encoding="utf-8") as f:
    list_path = f.name
    for p in files:
        f.write(f"file '{p.as_posix()}'\nduration {1/FPS}\n")
    f.write(f"file '{files[-1].as_posix()}'\n")  # last frame needs to be repeated

r = subprocess.run(
    ["ffmpeg","-y","-f","concat","-safe","0","-i", list_path,
     "-vf","format=gray16le,format=yuv420p","-r", str(FPS),
     "-c:v","libx264","-crf","18","-preset","medium", str(OUTPUT)],
    capture_output=True, text=True,
)
os.unlink(list_path)
print("returncode:", r.returncode)
print(r.stderr[-2000:])
```
